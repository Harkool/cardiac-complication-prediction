# 🧪 Exploratory Data Analysis for Cardiac Surgery Dataset

```
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

# 读取数据
df = pd.read_csv("../data/processed/cleaned_data.csv")
df.shape
```

## 📌 缺失值分布

```
missing = df.isnull().mean().sort_values(ascending=False)
missing[missing > 0].plot(kind='barh', figsize=(8, 10), title='Missing Value Ratio')
plt.tight_layout()
plt.show()
plt.savefig('../outputs/figures/missing_ratio.png')
```

## 🏷 标签分布（7d内术后肺部并发症）

```
target = '7d内术后肺部并发症(0=否，1=是)'
sns.countplot(data=df, x=target)
plt.title("Complication Label Distribution")
plt.show()
plt.savefig('../outputs/figures/label_distribution.png')
```

## 📊 数值变量分布示例

```
num_cols = df.select_dtypes(include=[np.number]).columns
df[num_cols[:6]].hist(bins=30, figsize=(14, 10))
plt.suptitle("Distributions of Selected Numerical Features")
plt.tight_layout()
plt.show()
plt.savefig('../outputs/figures/numeric_distribution.png')
```

## 🔍 标签与特征的相关性

```
corr = df.corr(numeric_only=True)[target].drop(target).sort_values(ascending=False)
corr.head(20).plot(kind='barh', title='Top Features Correlated with Complication')
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
plt.savefig('../outputs/figures/top_correlations.png')
```

## 🎯 分组对比：是否干预 vs 并发症率

```
if 'Intervention' in df.columns:
    pd.crosstab(df['Intervention'], df[target]).apply(lambda r: r / r.sum(), axis=1).plot(kind='bar', stacked=True)
    plt.title("Intervention vs Complication")
    plt.ylabel("Proportion")
    plt.xticks([0,1], ['No Block', 'Nerve Block'], rotation=0)
    plt.tight_layout()
    plt.show()
plt.savefig('../outputs/figures/intervention_vs_complication.png')
```
